```python
import pandas as pd
```

# Load datasets

```python
road_df = pd.read_csv("/content/drive/MyDrive/COM /Road.csv")
```

```python
weather_df = pd.read_csv("/content/drive/MyDrive/COM /Weather_Fuel_merge.csv")
```

1. Standardize Weather date columns

```python
weather_df["Date"] = pd.to_datetime(weather_df["Month_Year"])
weather_df["Year"] = weather_df["Date"].dt.year
weather_df["Month"] = weather_df["Date"].dt.month
weather_df["Month_Year_Match"] = weather_df["Date"].dt.strftime("%Y-%m")
```

2. Extract unique periods from the weather dataset

```python
time_periods = weather_df[
    ["Year", "Month", "Month_Year_Match"]
].drop_duplicates()
```

3. Drop existing static date columns from road dataset

```python
road_features = road_df.drop(
    columns=["Month", "Year", "Month_Year"], errors="ignore"
)
```

4. Cross join road segments with every time period

```python
road_expanded = road_features.merge(time_periods, how="cross")
```

5. Merge expanded road data with weather data

```python
final_merged = pd.merge(
    road_expanded,
    weather_df,
    left_on=["Year", "Month"],
    right_on=["Year", "Month"],
    how="left",
)
```

```python
import pandas as pd

weather_df["Date"] = pd.to_datetime(weather_df["Month_Year"])

# Filter weather dataset for September 2019
weather_sep2019 = weather_df[
    (weather_df["Date"].dt.year == 2019) & (weather_df["Date"].dt.month == 9)
]

# Merge on matching Month and Year
merged_df = pd.merge(road_df, weather_sep2019, on=["Year", "Month"], how="inner")
```
